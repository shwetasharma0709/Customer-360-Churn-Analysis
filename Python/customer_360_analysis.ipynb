```python
churned_customers = (df["Churn"] == "Yes").sum()
total_customers = len(df)

churn_rate = (churned_customers / total_customers) * 100

print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Churn Rate:", round(churn_rate, 2), "%")
```

```python
plan_churn = pd.crosstab(
    df["Subscription_Plan"],
    df["Churn"],
    normalize="index"
) * 100

plan_churn
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=plan_churn.index,
    y=plan_churn["Yes"]
)

plt.title("Churn Rate by Subscription Plan")
plt.xlabel("Subscription Plan")
plt.ylabel("Churn Rate (%)")
plt.ylim(0, 25)

plt.show()
```

```python
df["Customer_Segment"] = pd.qcut(
    df["Total_Spend"],
    q=4,
    labels=["Low Value", "Medium Value", "High Value", "VIP"]
)

df["Customer_Segment"].value_counts().sort_index()
```

```python
segment_churn = pd.crosstab(
    df["Customer_Segment"],
    df["Churn"],
    normalize="index"
) * 100

segment_churn
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=segment_churn.index,
    y=segment_churn["Yes"]
)

plt.title("Churn Rate by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Churn Rate (%)")
plt.ylim(0, 25)

plt.show()
```

```python
numeric_cols = [
    "Age",
    "Tenure_Months",
    "Total_Orders",
    "Avg_Order_Value",
    "Total_Spend",
    "Support_Tickets",
    "Satisfaction_Score",
    "Days_Since_Last_Purchase"
]

correlation = df[numeric_cols].corrwith(
    df["Churn"].map({"No": 0, "Yes": 1})
).sort_values(ascending=False)

correlation
```

```python
df["Purchase_Risk"] = np.where(
    df["Days_Since_Last_Purchase"] > 120, 1, 0
)

df["Support_Risk"] = np.where(
    df["Support_Tickets"] >= 3, 1, 0
)

df["Satisfaction_Risk"] = np.where(
    df["Satisfaction_Score"] < 3, 1, 0
)

df["Risk_Score"] = (
    df["Purchase_Risk"]
    + df["Support_Risk"]
    + df["Satisfaction_Risk"]
)

df[[
    "Customer_ID",
    "Days_Since_Last_Purchase",
    "Support_Tickets",
    "Satisfaction_Score",
    "Risk_Score",
    "Churn"
]].head(10)
```

```python
df["Risk_Level"] = pd.cut(
    df["Risk_Score"],
    bins=[-1, 0, 1, 2, 3],
    labels=["Low", "Medium", "High", "Critical"]
)

df["Risk_Level"].value_counts().sort_index()
```

```python
risk_churn = pd.crosstab(
    df["Risk_Level"],
    df["Churn"],
    normalize="index"
) * 100

risk_churn
```

```python
risk_churn_rate = risk_churn["Yes"]

plt.figure(figsize=(8, 5))

risk_churn_rate.plot(kind="bar")

plt.title("Churn Rate by Customer Risk Level")
plt.xlabel("Risk Level")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.ylim(0, 100)

plt.show()
```
